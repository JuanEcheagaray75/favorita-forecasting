```python
items = catalog.load("items_processed")
calendar = catalog.load("calendar")
stores = catalog.load("stores_processed")
train_processed = catalog.load("train_processed")


SATURDAY_ID = 7
calendar_processed = (
    calendar.withColumn("year", F.year("date"))
    .withColumn("week", F.weekofyear("date"))
    .filter(F.dayofweek("date") == SATURDAY_ID)
    .withColumn("time_percentage", F.percent_rank().over(W.orderBy("date")))
)

weekly_sales = (
    train_processed.withColumn("year", F.year("date"))
    .withColumn("week", F.weekofyear("date"))
    .groupby("year", "week", "store_id", "item_id")
    .agg(F.sum("unit_sales").alias("unit_sale"))
)

store_item_window = W.partitionBy("store_id", "item_id")
city_item_window = W.partitionBy("date", "city", "item_id")
real_purchase = F.when(F.col("unit_sale").isNotNull(), F.col("date"))

is_product_available = F.when(
    F.first("unit_sale", ignorenulls=True).over(city_item_window).isNotNull(), True
).otherwise(False)

all_sales = (
    calendar_processed.crossJoin(items)
    .crossJoin(stores)
    .join(other=weekly_sales, on=["year", "week", "store_id", "item_id"], how="left")
    .withColumn(
        "first_product_store_purchase", F.min(real_purchase).over(store_item_window)
    )
    .withColumn("is_product_available", is_product_available)
    .filter(F.col("date") >= F.col("first_product_store_purchase"))
    .fillna(value=0, subset=["unit_sale"])
    .withColumn("unit_sale", F.greatest(F.col("unit_sale"), F.lit(0.0)))
    .drop("year", "week")
)
```

```python
MAX_SALE_LAG = 6
lags = list(range(1, MAX_SALE_LAG + 1))

store_item_window = W.partitionBy("store_id", "item_id").orderBy("date")
store_window = W.partitionBy("store_id").orderBy("date")
item_window = W.partitionBy("item_id").orderBy("date")
class_window = W.partitionBy("class").orderBy("date")
family_window = W.partitionBy("family").orderBy("date")
city_window = W.partitionBy("city").orderBy("date")
state_window = W.partitionBy("state").orderBy("date")

# Window sizes
period_sizes = [4, 8, 12]


def add_lags(
    df: DataFrame, lag_col: str, window: WindowSpec, lags: list[int]
) -> DataFrame:
    lag_cols = {
        f"lag_{lag_col}_{i}": F.lag(col=lag_col, offset=i).over(window) for i in lags
    }

    return df.withColumns(lag_cols)


def add_agg_over_windows(
    df: DataFrame,
    val_col: str,
    agg_spark_fn: Callable,
    prefix: str,
    window: WindowSpec,
    periods: list[int],
) -> DataFrame:
    mean_cols = {
        f"{prefix}_{i}": agg_spark_fn(val_col).over(window.rowsBetween(-i, -1))
        for i in periods
    }

    return df.withColumns(mean_cols)


def count_non_zero_wrap(col_name: str):
    return F.count(F.when(F.col(col_name) > 0, True).otherwise(False))


def mean_real_purchase(col_name):
    return F.mean(F.when(F.col(col_name) > 0, F.col(col_name)))


def add_smooth_rbf(
    df: DataFrame,
    peak_at: int,
    alpha: float,
    date_col: str = "date",
    period_length: int = 52,
    start_date: str = "2013-01-01",
) -> DataFrame:
    t = F.ceil(F.date_diff(F.col(date_col), F.to_date(F.lit(start_date))) / 7)
    smooth_rbf = F.exp(
        -(F.lit(period_length) ** 2)
        / F.lit(alpha)
        * F.sin(F.pi() * (t - F.lit(peak_at)) / F.lit(period_length)) ** 2
    )

    return df.withColumn(f"smooth_rbf_{peak_at}_{int(alpha)}", smooth_rbf)


def add_one_side_rbf(
    df: DataFrame, peak_at: int, alpha: float, date_col: str = "date"
) -> DataFrame:
    shrinkage = 52 / (peak_at * alpha**2)
    peak_rbf = F.exp(-shrinkage * (F.lit(peak_at) - F.weekofyear(date_col)) ** 2)

    return df.withColumn(f"one_side_rbf_{peak_at}_{int(alpha)}", peak_rbf)


df = (
    all_sales.transform(
        add_lags, lag_col="unit_sale", window=store_item_window, lags=lags
    )
    .withColumn("sales_diff_week", F.col("lag_unit_sale_1") - F.col("lag_unit_sale_2"))
    # Periodic Exponential Peaks
    .transform(add_one_side_rbf, peak_at=52, alpha=10)
    .transform(add_one_side_rbf, peak_at=18, alpha=10)
    .transform(add_one_side_rbf, peak_at=40, alpha=10)
    # Smooth week effects
    .transform(add_smooth_rbf, peak_at=52, alpha=15)
    .transform(add_smooth_rbf, peak_at=40, alpha=15)
    .transform(add_smooth_rbf, peak_at=18, alpha=15)
    # Smooth month effects
    .transform(add_smooth_rbf, peak_at=50, alpha=75)
    .transform(add_smooth_rbf, peak_at=36, alpha=75)
    .transform(add_smooth_rbf, peak_at=10, alpha=75)
    .transform(
        add_agg_over_windows,
        val_col="sales_diff_week",
        prefix="sales_diff_mean",
        agg_spark_fn=F.mean,
        window=W.partitionBy("store_id", "item_id").orderBy("date"),
        periods=period_sizes,
    )
    .transform(
        add_agg_over_windows,
        val_col="sales_diff_week",
        prefix="sales_diff_std",
        agg_spark_fn=F.std,
        window=W.partitionBy("store_id", "item_id").orderBy("date"),
        periods=period_sizes,
    )
    .transform(
        add_agg_over_windows,
        val_col="unit_sale",
        prefix="store_item_real_mean",
        agg_spark_fn=mean_real_purchase,
        window=W.partitionBy("store_id", "item_id").orderBy("date"),
        periods=period_sizes,
    )
    .transform(
        add_agg_over_windows,
        val_col="unit_sale",
        prefix="store_item_max",
        agg_spark_fn=F.max,
        window=W.partitionBy("store_id", "item_id").orderBy("date"),
        periods=[4],
    )
    .transform(
        add_agg_over_windows,
        val_col="unit_sale",
        prefix="store_item_std",
        agg_spark_fn=F.stddev_samp,
        window=W.partitionBy("store_id", "item_id").orderBy("date"),
        periods=period_sizes,
    )
    .transform(
        add_agg_over_windows,
        val_col="unit_sale",
        prefix="family_mean",
        agg_spark_fn=F.mean,
        window=W.partitionBy("family").orderBy("date"),
        periods=period_sizes,
    )
    .transform(
        add_agg_over_windows,
        val_col="unit_sale",
        prefix="class_mean",
        agg_spark_fn=F.mean,
        window=W.partitionBy("class").orderBy("date"),
        periods=period_sizes,
    )
    .transform(
        add_agg_over_windows,
        val_col="unit_sale",
        prefix="class_real_mean",
        agg_spark_fn=mean_real_purchase,
        window=W.partitionBy("class").orderBy("date"),
        periods=period_sizes,
    )
    .transform(
        add_agg_over_windows,
        val_col="unit_sale",
        prefix="class_std",
        agg_spark_fn=F.stddev_samp,
        window=W.partitionBy("class").orderBy("date"),
        periods=[4, 12],
    )
    .transform(
        add_agg_over_windows,
        val_col="unit_sale",
        prefix="cluster_mean",
        agg_spark_fn=F.mean,
        window=W.partitionBy("cluster").orderBy("date"),
        periods=period_sizes,
    )
)


target_col = "unit_sale"
exclude_cols = [
    "store_id",
    "item_id",
    "date",
    "time_percentage",
    "family",
    "class",
    "perishable",
    "city",
    "state",
    "type",
    "cluster",
    "item_weight",
    "first_product_store_purchase",
    "is_product_available",
] + [target_col, "log_unit_sale"]


categorical_columns = ["family", "class", "cluster"]
numerical_cols = [
    c for c in df.columns if (c not in exclude_cols) and (c not in categorical_columns)
]

df.columns
```

```python
import time


import mlflow
from mlflow.models.signature import infer_signature
from pyspark.ml.pipeline import Pipeline, PipelineModel
from pyspark.ml.feature import (
    VectorAssembler,
    SQLTransformer,
    StandardScaler,
)
from pyspark.ml.regression import LinearRegression, LinearRegressionModel
from xgboost.spark import SparkXGBRegressor, SparkXGBRegressorModel


def get_linear_importance(
    model: LinearRegressionModel, input_cols: list[str]
) -> pd.DataFrame:
    model_coeffs = model.coefficients.T
    feature_importance = pd.DataFrame(
        model_coeffs.reshape(-1, 1), columns=["importance"]
    )
    feature_importance["feature"] = input_cols
    feature_importance["abs_coeff"] = np.abs(feature_importance["importance"])
    feature_importance = feature_importance.sort_values(by="abs_coeff", ascending=True)

    return feature_importance


def get_xgboost_importance(
    model: SparkXGBRegressorModel, input_cols: list[str]
) -> pd.DataFrame:
    feature_importance = pd.DataFrame(
        model.get_feature_importances().items(),
        columns=["feature_idx", "importance"],
    )
    feature_importance["feature"] = input_cols
    feature_importance = feature_importance.sort_values(
        by="importance", ascending=False
    )
    return feature_importance


def plot_feature_importance(feature_importance: pd.DataFrame) -> Figure:
    fig, ax = plt.subplots(figsize=(9, 6))
    plt.barh(
        y=feature_importance["feature"],
        width=feature_importance["importance"],
    )
    ax.set_xlabel("Feature Importance")
    ax.set_ylabel("Variable")
    ax.set_title("Feature Importance")
    plt.tight_layout()
    plt.close(fig)
    return fig


def hexbin_real_vs_pred(
    df_preds: DataFrame,
    label_col: str,
    prediction_col: str,
    sampling_factor: float = 0.1,
) -> Figure:
    sampled_preds = (
        df_preds.select(label_col, prediction_col).sample(sampling_factor).toPandas()
    )

    fig, ax = plt.subplots(figsize=(9, 6))
    plt.hexbin(sampled_preds[prediction_col], sampled_preds[label_col])
    ax.set_xlabel("Prediction")
    ax.set_ylabel("Real")
    ax.set_title("Prediction vs Real Value")
    plt.tight_layout()
    plt.close(fig)
    return fig
```

```python
def rmsle(df: DataFrame, label_col: str, pred_col: str) -> DataFrame:
    return df.agg(
        F.sqrt(F.mean(F.pow(F.log1p(label_col) - F.log1p(pred_col), 2))).alias("rmsle")
    )


def mse(df: DataFrame, label_col: str, pred_col: str) -> DataFrame:
    return df.agg(F.mean(F.pow(F.col(label_col) - F.col(pred_col), 2)).alias("mse"))
```

```python
numerical_feature_assembler = VectorAssembler(
    inputCols=all_input_cols, outputCol="features", handleInvalid="keep"
)
base_model = SparkXGBRegressor(
    features_col="features",
    label_col=target_col,
    prediction_col="base_prediction",
    tree_method="hist",
    objective="reg:squaredlogerror",
)
fallback_prediction = SQLTransformer(
    statement="""
    SELECT
        *,
        lag_unit_sale_1 AS fallback_prediction
    FROM
        __THIS__
    """
)
negativity_base_check = SQLTransformer(
    statement="""
    SELECT
        *,
        GREATEST(0, base_prediction) AS corrected_base_prediction
    FROM
        __THIS__
    """
)
base_model_pipe = Pipeline(
    stages=[
        numerical_feature_assembler,
        base_model,
        fallback_prediction,
        negativity_base_check,
    ]
)

# Direct Loss Estimation
dle_target_pipe = Pipeline(
    stages=[
        SQLTransformer(
            statement=f"""
            SELECT
                *,
                POW(LOG1P(corrected_base_prediction) - LOG1P({target_col}), 2) AS base_msle
            FROM
                __THIS__
            """
        ),
        SQLTransformer(
            statement=f"""
            SELECT
                *,
                POW(LOG1P(fallback_prediction) - LOG1P({target_col}), 2) AS fallback_msle
            FROM
                __THIS__
            """
        ),
    ]
)

# Nanny Models
nanny_features_base = VectorAssembler(
    inputCols=[*numerical_cols, "corrected_base_prediction"],
    outputCol="nanny_base_features",
    handleInvalid="keep",
)
nanny_features_fallback = VectorAssembler(
    inputCols=[*numerical_cols, "fallback_prediction"],
    outputCol="nanny_fallback_features",
    handleInvalid="keep",
)
nanny_base_model = SparkXGBRegressor(
    features_col="nanny_base_features",
    label_col="base_msle",
    prediction_col="nanny_base_loss_estimate",
    objective="reg:squaredlogerror",
    tree_method="hist",
)
nanny_fallback_model = SparkXGBRegressor(
    features_col="nanny_fallback_features",
    label_col="fallback_msle",
    prediction_col="nanny_fallback_loss_estimate",
    objective="reg:squaredlogerror",
    tree_method="hist",
)
nanny_models_pipe = Pipeline(
    stages=[
        nanny_features_base,
        nanny_features_fallback,
        nanny_base_model,
        nanny_fallback_model,
    ]
)

base_model = base_model_pipe.fit(train)
base_train_preds = base_model.transform(train)
dle_model = dle_target_pipe.fit(base_train_preds)
train_with_dle = dle_model.transform(base_train_preds)
nanny_models = nanny_models_pipe.fit(train_with_dle)

naive_inference_selector = SQLTransformer(
    statement="""
    SELECT
        *,
        CASE
            WHEN nanny_base_loss_estimate < nanny_fallback_loss_estimate THEN corrected_base_prediction
            ELSE fallback_prediction
        END AS final_prediction
    FROM
        __THIS__
    """
)
# Final Check
negativity_check = SQLTransformer(
    statement="""
    SELECT
        *,
        GREATEST(final_prediction, 0) AS prediction
    FROM
        __THIS__
    """
)

final_pipe = Pipeline(
    stages=[base_model, nanny_models, naive_inference_selector, negativity_check]
)
final_model = final_pipe.fit(train)

final_train_preds = final_model.transform(train)
final_val_preds = final_model.transform(val)

train_rmsle = rmsle(
    final_train_preds, label_col="unit_sale", pred_col="prediction"
).collect()[0][0]
val_rmsle = rmsle(
    final_val_preds, label_col="unit_sale", pred_col="prediction"
).collect()[0][0]

print(f"Train RMSLE: {train_rmsle}")
print(f"Val RMSLE: {val_rmsle}")
```

```python
experiment = mlflow.set_experiment(experiment_name="favorita-feature-engineer")


mlflow.pyspark.ml.autolog(log_models=False, log_datasets=False)
with mlflow.start_run(experiment_id=experiment.experiment_id):
    numerical_feature_assembler = VectorAssembler(
        inputCols=all_input_cols, outputCol="features", handleInvalid="keep"
    )
    numerical_scaler = StandardScaler(
        inputCol="features",
        outputCol="scaled_features",
        withMean=True,
        withStd=True,
    )
    base_pred_model = LinearRegression(
        featuresCol="scaled_features",
        labelCol=target_col,
        predictionCol="original_prediction",
        regParam=1,
        elasticNetParam=0,
        standardization=False,
    )

    # Final Check
    negativity_check = SQLTransformer(
        statement="""
                SELECT
                    *,
                    CASE
                        WHEN original_prediction < 0 THEN 0
                        ELSE original_prediction
                    END AS prediction
                FROM __THIS__"""
    )

    pipe = Pipeline(
        stages=[
            numerical_feature_assembler,
            numerical_scaler,
            base_pred_model,
            negativity_check,
        ]
    )

    train_start = time.time()
    model = pipe.fit(train)
    train_end = time.time()

    train_preds = model.transform(train)
    val_preds = model.transform(val)

    # Base performance metrics
    train_rmsle = rmsle(
        train_preds, label_col="unit_sale", pred_col="prediction"
    ).collect()[0][0]

    val_rmsle = rmsle(
        val_preds, label_col="unit_sale", pred_col="prediction"
    ).collect()[0][0]
    train_rows = train.count()
    val_rows = val.count()
    train_time = train_end - train_start

    input_example = train.select(*all_input_cols, target_col).limit(5)
    pd_sample = input_example.toPandas()
    y = pd_sample.pop(target_col).values
    X = pd_sample
    signature = infer_signature(model_input=X, model_output=y)

    mlflow.spark.log_model(
        spark_model=model,
        artifact_path="model",
        signature=signature,
        input_example=pd_sample,
    )
    mlflow.log_metrics(
        {
            "train_rmsle": train_rmsle,
            "val_rmsle": val_rmsle,
            "train_rows": train_rows,
            "val_rows": val_rows,
            "train_time": train_time,
        }
    )

    feature_importance = get_linear_importance(model.stages[-2], all_input_cols)
    feat_import_plot = plot_feature_importance(feature_importance)
    mlflow.log_figure(feat_import_plot, "fig/feature_importance.png")
```
