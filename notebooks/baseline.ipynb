```python
df: DataFrame = catalog.load("weekly_sales_filled")
calendar: DataFrame = catalog.load("calendar").withColumn(
    "percent_rank", F.percent_rank().over(Window.orderBy("date"))
)

# Suppose we do a train test split with 80, 15, 5
train_perc = 0.8
val_perc = 0.15
test_perc = 1 - train_perc + val_perc
cutoffs = (
    calendar.withColumn(
        "idx",
        F.when(F.col("percent_rank") < train_perc, "train")
        .when(F.col("percent_rank") < train_perc + val_perc, "val")
        .otherwise("test"),
    )
    .groupby("idx")
    .agg(F.max("date").alias("cutoff"))
    .toPandas()
    .set_index("idx")["cutoff"]
    .to_dict()
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt


train = df.filter(F.col("date") <= cutoffs["train"])

# Data Exploration
mean_sales = (
    train.groupby("date")
    .agg(
        F.mean("unit_sale").alias("mean_sale"),
        F.percentile_approx("unit_sale", 0.05).alias("5_perc_unit_sale"),
        F.percentile_approx("unit_sale", 0.95).alias("95_perc_unit_sale"),
    )
    .sort(F.asc("date"))
    .toPandas()
)
```

```python
store_item_wspec = Window.partitionBy("store_id", "item_id").orderBy("date")
store_wspec = Window.partitionBy("store_id", "date").orderBy(F.asc("total_item_sale"))


def add_product_rank(
    df: DataFrame, rolled_sales_col: str, output_col: str = "product_rank"
) -> DataFrame:
    return df.withColumn(
        output_col,
        F.percent_rank().over(
            Window.partitionBy("store_id", "date").orderBy(F.asc(rolled_sales_col))
        ),
    )


def add_is_low_rank(
    df: DataFrame,
    product_rank_col: str,
    rank_thresh: float = 0.1,
    output_col: str = "is_low_rank",
) -> DataFrame:
    return df.withColumn(
        output_col, F.when(F.col(product_rank_col) < rank_thresh, 1).otherwise(0)
    )


product_ranking = (
    train.withColumn(
        "roll_4_item_sale",
        F.sum("unit_sale").over(store_item_wspec.rowsBetween(-4, -1)),
    )
    .transform(
        add_product_rank,
        rolled_sales_col="roll_4_item_sale",
        output_col="product_rank_4_weeks",
    )
    .transform(
        add_is_low_rank,
        product_rank_col="product_rank_4_weeks",
        rank_thresh=0.119532,
        output_col="is_low_rank_in_4_weeks",
    )
    .withColumn("is_zero_sale", F.when(F.col("unit_sale") == 0, 1).otherwise(0))
)

# Check if there are some useful threhsolds
rank_zero_thresh = product_ranking.groupby("is_zero_sale").agg(
    F.mean("product_rank_4_weeks").alias("mean_rank")
)
```

```python
from sklearn.metrics import matthews_corrcoef


def convert_to_int(col: str):
    return F.when(F.col(col) == 1, 1).otherwise(-1)


sample = (
    product_ranking.sample(0.1)
    .select(
        convert_to_int("is_zero_sale").alias("is_zero_sale"),
        convert_to_int("is_low_rank_in_4_weeks").alias("is_low_rank"),
    )
    .limit(200_000)
    .toPandas()
)

matthews_corrcoef(sample["is_zero_sale"], sample["is_low_rank"])
```

```python
num_bins = 30


def compute_histogram(df, bins, value_col):
    return (
        df.select(F.col(value_col))
        .withColumn(
            "bin",
            F.expr(f"""
            CASE
                { " ".join([f"WHEN {value_col} < {bins[i+1]} THEN {i}" for i in range(len(bins) - 1)]) }
                ELSE {len(bins) - 1}
            END
        """),
        )
        .groupBy("bin")
        .count()
        .orderBy("bin")
    )


product_rank_mistake = product_ranking.filter(F.col("is_zero_sale") == 1).filter(
    F.col("is_low_rank_in_4_weeks") == 0
)

product_rank_mistake_dist = product_rank_mistake.approxQuantile(
    "product_rank_4_weeks",
    probabilities=[i / num_bins for i in range(num_bins)],
    relativeError=0.001,
)

hist_ranks = compute_histogram(
    product_rank_mistake,
    bins=product_rank_mistake_dist,
    value_col="product_rank_4_weeks",
)
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.getActiveSession()

bins_df = spark.createDataFrame(
    [
        (
            i,
            float(product_rank_mistake_dist[i]),
            float(product_rank_mistake_dist[i + 1]),
        )
        for i in range(len(product_rank_mistake_dist) - 1)
    ],
    ["bin", "bin_start", "bin_end"],
)
```

```python
histogram = (
    hist_ranks.join(other=bins_df, on="bin")
    .select("bin_end", "count")
    .sort(F.asc("bin_end"))
    .toPandas()
)

plt.figure(figsize=(10, 6))
plt.bar(histogram["bin_end"], histogram["count"], align="edge")
plt.xlabel("Product Rank")
plt.ylabel("Count")
plt.title("Product Rank Distribution")
plt.show()
```

```python
# Add counts of real sales over last 4 weeks
# Add counts of zero purchases

product_ranking.withColumn(
    "mean_ranking",
    F.mean("product_rank_4_weeks").over(store_item_wspec.rowsBetween(-8, -1)),
).withColumn(
    "cumprod_ranking",
    F.product("product_rank_4_weeks").over(store_item_wspec.rowsBetween(-8, -1)),
).withColumn(
    "product_ranking_variation",
    F.std("product_rank_4_weeks").over(store_item_wspec.rowsBetween(-4, -1)),
).withColumn(
    "product_rank_cv", F.col("product_ranking_variation") / F.col("mean_ranking")
)
```

## Finding Zero Sale Streaks

```python
historic_purchase_proba = (
    product_ranking.groupby("date")
    .agg((1 - F.mean("is_zero_sale")).alias("purchase_probability"))
    .sort(F.asc("date"))
    .toPandas()
)

historic_purchase_proba["date"] = pd.to_datetime(historic_purchase_proba["date"])

plt.figure(figsize=(10, 6))
plt.plot(
    historic_purchase_proba["date"], historic_purchase_proba["purchase_probability"]
)
plt.xlabel("Date")
plt.ylabel("Purchase Probability")
plt.show()
```

```python
# Build indicators for a sale being 0
store_item_wspec = Window.partitionBy("store_id", "item_id").orderBy("date")

zero_sale_eda = train.withColumn(
    "is_sale_zero", F.when(F.col("unit_sale") == 0, True).otherwise(False)
).withColumn("lag_1", F.lag("unit_sale").over(store_item_wspec))

sample_store = 1
sample_item = 364834

zero_example = (
    zero_sale_eda.filter(
        (F.col("store_id") == sample_store) & (F.col("item_id") == sample_item)
    )
    .sort(F.asc("date"))
    .select("date", "unit_sale", "is_sale_zero")
    .toPandas()
)

zero_example["date"] = pd.to_datetime(zero_example["date"])
plt.figure(figsize=(10, 6))
plt.plot(zero_example["date"], zero_example["unit_sale"])
plt.show()
```

## Forecasting actual sale

```python
from pyspark.ml.pipeline import Pipeline
from pyspark.ml.evaluation import (
    RegressionEvaluator,
    BinaryClassificationEvaluator,
    MulticlassClassificationEvaluator,
)
from pyspark.ml.feature import StringIndexer, VectorAssembler
from pyspark.ml.regression import DecisionTreeRegressor
from pyspark.ml.classification import DecisionTreeClassifier


f1_evaluator = MulticlassClassificationEvaluator(metricLabel=1, beta=1.0)

propensity_pipe = Pipeline(
    stages=[
        StringIndexer(
            inputCols=["family", "class", "city", "type", "cluster"],
            outputCols=[
                "family_idx",
                "city_idx",
                "type_idx",
                "cluster_idx",
            ],
        ),
        VectorAssembler(
            inputCols=[
                "roll_4_item_sale",
                "product_rank_4_weeks",
                "is_low_rank_in_4_weeks",
                "family_idx",
                "class_idx",
                "city_idx",
                "type_idx",
                "cluster_idx",
            ],
            outputCol="features",
            handleInvalid="skip",
        ),
        DecisionTreeClassifier(featuresCol="features", seed=0),
    ]
)

train_propensity = product_ranking.withColumn(
    "label", F.when(F.col("unit_sale") > 0, 1.0).otherwise(0.0)
)

propensity_model = propensity_pipe.fit(train_propensity)

f1_evaluator.evaluate(
    propensity_model.transform(train_propensity).sample(0.1).limit(100_000)
)
```
